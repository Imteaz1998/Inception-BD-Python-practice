# How object access attributes in OOP

```python
class Person:
    def __init__(self,name,country):
        self.name=name
        self.country=country
    def greetings(self):
        print(f"Hello {self.name}, Welcome to {self.country}")
```

```python
p=Person("Alex","Banglades")
```

```python
p.greetings()
```

# Pass by Reference

```python
class Person:
    def __init__(self,name,country):
        self.name=name
        self.country=country

def greetings(person):
    print(f"Hi my name is {person.name} and I am from {person.country}")
    p1=Person("Alex","India")
    return p1
```

```python
p=Person("Imteaz","Malta")
a=greetings(p)
a.name,a.country
```

Inside greetings():

It prints the name & country of the person you pass (Imteaz, Malta).

BUT THEN you create a new Person object:

p1 = Person("Alex","India")


And you return this new object, not the original.

# Encapsulation

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.__age = age # Private property

p1 = Person("Emil", 25)
print(p1.name)
print(p1.__age) # Show error
```
